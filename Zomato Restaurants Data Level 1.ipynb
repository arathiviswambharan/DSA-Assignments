# TASK 1
Data Exploration and Preprocessing

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# 1. Load the dataset

data = pd.read_csv (r"C:\Users\arund\Downloads\Dataset  (1).csv")

data.head ()
```

```python
# 2. Number of rows and columns

print("Number of rows and columns:", data.shape)
```

```python
# 3. Check missing values in each column

print("\nMissing values:\n", data.isnull().sum())
```

```python
# Create a base map centered around the mean location
center_lat = data["Latitude"].mean()
center_lon = data["Longitude"].mean()

m = folium.Map(location=[center_lat, center_lon], zoom_start=12)

# Add markers
for _, row in data.iterrows():
    folium.Marker(
        location=[row["Latitude"], row["Longitude"]],
        popup=row.get("Restaurant Name", "Restaurant"),
        icon=folium.Icon(color="red", icon="cutlery", prefix="fa")
    ).add_to(m)

display(m)
```

```python
# Average rating by country
country_rating = data.groupby("Country Code")["Aggregate rating"].mean()

plt.figure(figsize=(10,6))
country_rating.plot(kind='bar', color='green')
plt.title("Average Restaurant Rating by Country")
plt.xlabel("Country Code")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 4. FIND STATISTICAL CORRELATION
# ----------------------------
# Does location correlate with rating?
corr_city = data.groupby("City")["Aggregate rating"].mean().corr(
            data.groupby("City").size())

corr_country = data.groupby("Country Code")["Aggregate rating"].mean().corr(
               data.groupby("Country Code").size())

print("Correlation between city restaurant count and average rating:", corr_city)
print("Correlation between country restaurant count and average rating:", corr_country)
```
