```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# 1. Load the dataset

data = pd.read_csv (r"C:\Users\arund\Downloads\Dataset  (1).csv")

data.head ()
```

```python
# Total number of restaurants
total = len(data)

# Percentage offering table booking
table_booking_percent = (data['Has Table booking'].str.lower() == "yes").mean() * 100

# Percentage offering online delivery
online_delivery_percent = (data['Has Online delivery'].str.lower() == "yes").mean() * 100

print(f"Percentage of restaurants offering table booking: {table_booking_percent:.2f}%")
print(f"Percentage of restaurants offering online delivery: {online_delivery_percent:.2f}%")
```

```python
# Clean column values
data['Has Table booking'] = data['Has Table booking'].str.strip().str.lower()

# Average rating for restaurants WITH table booking
avg_with_booking = data[data['Has Table booking'] == 'yes']['Aggregate rating'].mean()

# Average rating for restaurants WITHOUT table booking
avg_without_booking = data[data['Has Table booking'] == 'no']['Aggregate rating'].mean()

print(f"Average rating (with table booking): {avg_with_booking:.2f}")
print(f"Average rating (without table booking): {avg_without_booking:.2f}")
```

```python
# Standardize values
data['Has Online delivery'] = data['Has Online delivery'].str.strip().str.lower()

# Group by price range and calculate percentage offering online delivery
online_delivery_by_price = (
    data.groupby("Price range")["Has Online delivery"]
      .apply(lambda x: (x == "yes").mean() * 100)
      .reset_index()
)

online_delivery_by_price.columns = ["Price Range", "Online Delivery (%)"]

print("Percentage of restaurants offering online delivery by price range:")
print(online_delivery_by_price)
```

# Task 2
Price Range Analysis

```python
# Find the most common price range
most_common_price = data['Price range'].mode()[0]

print(f"The most common price range is: {most_common_price}")
```

```python
# 1. AVERAGE RATING BY PRICE RANGE

avg_rating_by_price = (
    data.groupby("Price range")["Aggregate rating"]
    .mean()
    .reset_index()
)

print("Average Rating for Each Price Range:")
print(avg_rating_by_price)
```

```python
# 2. IDENTIFY COLOR OF HIGHEST AVERAGE RATING

# Find price range with highest average rating
best_price_range = avg_rating_by_price.loc[
    avg_rating_by_price["Aggregate rating"].idxmax(), "Price range"
]

# Get all rows with this price range
subset = data[data["Price range"] == best_price_range]

# Find the most common rating color for this price range
most_common_color = subset["Rating color"].mode()[0]

print(f"\nPrice range with highest average rating: {best_price_range}")
print(f"Color representing this price range: {most_common_color}")
```

# Task 3
Feature Engineering

```python
# 1. LENGTH OF RESTAURANT NAME

data["Restaurant_Name_Length"] = data["Restaurant Name"].astype(str).apply(len)


# 2. LENGTH OF ADDRESS

data["Address_Length"] = data["Address"].astype(str).apply(len)


# 3. WORD COUNT IN RESTAURANT NAME

data["Restaurant_Name_WordCount"] = data["Restaurant Name"].astype(str).apply(lambda x: len(x.split()))


# 4. WORD COUNT IN ADDRESS

data["Address_WordCount"] = data["Address"].astype(str).apply(lambda x: len(x.split()))


# 5. NUMBER OF CUISINES OFFERED

data["Cuisine_Count"] = data["Cuisines"].astype(str).apply(lambda x: len(x.split(", ")))


# 6. DOES THE RESTAURANT NAME CONTAIN SPECIAL CHARACTERS?

data["Has_SpecialChar"] = data["Restaurant Name"].astype(str).apply(
    lambda x: any(not char.isalnum() and not char.isspace() for char in x)
)

# Show first few rows with new features
data.head()
```

```python
# Standardize values to lowercase (important!)
data["Has Table booking"] = data["Has Table booking"].str.lower().str.strip()
data["Has Online delivery"] = data["Has Online delivery"].str.lower().str.strip()

# Encode Yes/No values into binary (1 or 0)
data["Table_Booking_Flag"] = data["Has Table booking"].map({"yes": 1, "no": 0})
data["Online_Delivery_Flag"] = data["Has Online delivery"].map({"yes": 1, "no": 0})

# Show first few rows
print(data[["Has Table booking", "Table_Booking_Flag",
          "Has Online delivery", "Online_Delivery_Flag"]].head())
```
